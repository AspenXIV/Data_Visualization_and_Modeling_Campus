## The reproducible analysis workflow

1. **Question:** What do we want to learn?
2. **Data:** Which observations and features can help answer it?
3. **Evidence:** What should we calculate and display to answer the question?
4. **Check:** Does the result make sense in context? Do a few original records support the meaning of the calculation?
5. **Conclusion:** What claim is supported?
6. **Limitation:** What should we avoid concluding?

```python
print("Plants above 6 cm:", sepal_over_6_count)
```

```python
print("Total plants:", len(iris_df["sepal_over_6_cm"]))
```

```python
print("Percent above 6 cm:", round(sepal_over_6_percent, 2))
```

### Step 1: Question

State a specific question that your dataset can answer. Define any important groups or comparisons.


**What is your research question 3?**

Your answer: What percent of plants have a petal length greater than, less than, or equal to their sepal width? I will compare the two measurements within each row. Equal means exactly equal.


### Prediction — before calculating


**What do you predict the result will be, and why?**

Your answer: I expect the greater-than group to be the largest and exact ties to be uncommon. These measure different parts of the flower, so I do not expect them to match often.


### Step 2: Data


Reuse the DataFrame and initial inspection from Investigation 1. Identify the observations and features relevant to this question; inspect additional features only if needed.


**Which observations and features will you use? Explain relevant units, any filtering, and how you will handle missing or not-applicable values using the dictionary. For a percentage, define the denominator.**

Your answer: I will use `petal_length_cm` and `sepal_width_cm` for all 150 plants. Both are measured in centimeters and have no missing values. Each plant will be counted once in Greater, Less, or Equal. Each percentage uses all 150 plants as its denominator. No rows will be removed.


### Step 3: Evidence

Write the code that answers your question. Break the analysis into separate cells for individual operations, as in the lecture, and add cells as needed. Display intermediate results where useful and the final result that answers the question. Explain any new calculated features. You may display a grouped summary directly without assigning it to a variable.


```python
petal_greater = iris_df["petal_length_cm"] > iris_df["sepal_width_cm"]
```

```python
petal_less = iris_df["petal_length_cm"] < iris_df["sepal_width_cm"]
```

```python
petal_equal = iris_df["petal_length_cm"] == iris_df["sepal_width_cm"]
```

```python
iris_df.loc[petal_greater, "petal_sepal_group"] = "Greater"
```

```python
iris_df.loc[petal_less, "petal_sepal_group"] = "Less"
```

```python
iris_df.loc[petal_equal, "petal_sepal_group"] = "Equal"
```

```python
petal_sepal_summary_df = iris_df["petal_sepal_group"].value_counts().to_frame("plant_count")
```

```python
petal_sepal_summary_df["percentage"] = petal_sepal_summary_df["plant_count"] / len(iris_df) * 100
```

```python
petal_sepal_summary_df.round(2)
```

```python
print("Exact ties:", petal_equal.sum())
```

**Which specific value or comparison answers your question? Explain how your calculations connect the selected data to that result.**

Your answer: Petal length is greater than sepal width in 100 plants, or 66.67%. It is less in 50 plants, or 33.33%. There are 0 exact ties, or 0%. The code compares the two measurements in each row, labels the result, and divides each group's count by all 150 plants. A separate count shows that no rows meet the equality condition.
